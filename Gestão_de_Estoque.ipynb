```python
def menu_saida():

  if menu_saida():
    while True:
        print("\n--- Sistema de Estoque ---")
        print("1 - Cadastrar produto")
        print("2 - Excluir produto")
        print("3 - Visualizar produtos")
        print("4 - Lançar entrada de produto")
        print("5 - Baixar saída de produto")
        print("6 - Produtos que estão com estoque abaixo de 5")
        print("0 - Sair")
        opcao = input("Escolha uma opção: ")

        if opcao == "1":
            # Lógica de cadastrar_produto() movida para cá
            nome = input("Nome do produto: ")
            if nome in estoque:
                print("Produto já cadastrado.")
            else:
                try:
                    quantidade = int(input("Quantidade inicial: "))
                    valor_unitario = float(input("Valor unitario: "))
                    estoque[nome] = {
                        "quantidade": quantidade,
                        "valor_unitario": valor_unitario
                    }
                    print(f"Produto '{nome}' cadastrado com sucesso.")
                except ValueError:
                    print("Entrada inválida para quantidade ou valor.")

        elif opcao == "2":
            # Lógica de excluir_produto() movida para cá
            nome = input("Nome do produto a excluir: ")
            if nome in estoque:
                del estoque[nome]
                print(f"Produto '{nome}' excluído do estoque.")
            else:
                print("Produto não encontrado.")

        elif opcao == "3":
            # Lógica de visualizar_produtos() movida para cá
            if not estoque:
                print("Estoque vazio.")
            else:
                print("Produtos em estoque:")
                for nome, dados in estoque.items():
                    valor_total = dados["quantidade"] * dados["valor_unitario"]
                    print(f"- {nome}: {dados['quantidade']} unidades | R$ {dados['valor_unitario']:.2f} cada | Total: R$ {valor_total:.2f}")

        elif opcao == "4":
            # Lógica de lancar_produto() movida para cá
            nome = input("Nome do produto para lançar entrada: ")
            if nome in estoque:
                try:
                    quantidade = int(input("Quantidade a adicionar: "))
                    estoque[nome]["quantidade"] += quantidade
                    print(f"{quantidade} unidades adicionadas ao produto '{nome}'.")
                except ValueError:
                    print("Entrada inválida para quantidade.")
            else:
                print("Produto não encontrado.")

        elif opcao == "5":
            # Lógica de baixar_produto() movida para cá
            nome = input("Nome do produto para baixar saída: ")
            if nome in estoque:
                try:
                    quantidade = int(input("Quantidade a baixar: "))
                    if quantidade > estoque[nome]["quantidade"]:
                        print("Quantidade insuficiente em estoque.")
                    else:
                        estoque[nome]["quantidade"] -= quantidade
                        print(f"{quantidade} unidades baixadas do produto '{nome}'.")
                except ValueError:
                    print("Entrada inválida para quantidade.")
            else:
                print("Produto não encontrado.")

        elif opcao == "6":
            # Lógica de produtos abaixo de 5 movida para cá
            if not estoque: # Verifica se o estoque está vazio primeiro
                print("Estoque vazio.")
            else:
                print("Produtos com estoque abaixo de 5 unidades:")
                encontrados = False # Flag para verificar se algum produto foi encontrado
                for nome, dados in estoque.items():
                    if dados["quantidade"] < 5:
                        valor_total = dados["quantidade"] * dados["valor_unitario"]
                        print(f"- {nome}: {dados['quantidade']} unidades | R$ {dados['valor_unitario']:.2f} cada | Total: R$ {valor_total:.2f}")
                        encontrados = True
                if not encontrados:
                    print("Nenhum produto com estoque abaixo de 5 unidades.")

        elif opcao == "0":
            print("Saindo...")
            break
        else:
            print("Opção inválida, tente novamente.")
```

```python
estoque_atual = {"Audi R8":13,  "Getta GLI": 10, "BMW 328i":9}
print(estoque_atual)
def adicionar_estoque(estoque_atual, produto, quantidade):
    if produto in estoque_atual:
        estoque_atual[produto] += quantidade
    else:
        estoque_atual[produto] = quantidade
    return estoque_atual

adicionar_estoque(estoque_atual, "Jaguar", 2)
print(estoque_atual)
```
