```python
import matplotlib.pyplot as plt

print(" FLUXO DE CAIXA")
dias_da_semana = ["Domingo","Segunda", "Terça", "Quarta", "Quinta", "Sexta", "Sábado"]
dias_solicitados = 0 #Inicia-se com 0 para continuar no loop
dia_atual = 1
saldo_total = 0
saldos_diarios = []
saldos_totais = []

while not (1 <= dias_solicitados <= 7):  # Repetir até que um número válido de dias seja inserido".
    dias_solicitados = int(input(" O saldo total de quantos dias da semana? (escolha entre 1 a 7) "))
    if not (1 <= dias_solicitados <= 7):
        print("Número Incorreto, tente novamente.")

while dia_atual <= dias_solicitados:
    nome_dia = dias_da_semana[dia_atual - 1]
    receita = float(input("Digite o depósito de {}: ".format(nome_dia)))
    despesa = float(input("Digite o valor a ser sacado de {}: ".format(nome_dia)))
    saldo_dia = receita - despesa
    saldo_total = saldo_total + saldo_dia
    saldos_diarios.append(saldo_dia)
    saldos_totais.append(saldo_total)
    print("Saldo de {}: R$ {:.2f}".format(nome_dia, saldo_dia))
    print("Saldo Total: R$ {:.2f}".format(saldo_total))
    dia_atual += 1

print("Verificação finalizada após {} dias.".format(dia_atual - 1))
print("SALDO TOTAL OU DA SEMANA: R$ {:.2f}".format(saldo_total))

# Gerar o gráfico
dias_para_plot = range(1, dias_solicitados + 1)

plt.figure(figsize=(10, 6))
plt.plot(dias_para_plot, saldos_diarios, label='Saldo Diário', marker='o')
plt.plot(dias_para_plot, saldos_totais, label='Saldo Total Acumulado', marker='o')
plt.xlabel('Dia')
plt.ylabel('Saldo (R$)')
plt.title('Saldo Diário e Total ao Longo do Tempo')
plt.grid(True)
plt.legend()
plt.show()
```
