```python
import math # Importa a biblioteca "math" para a função do logaritmo ser executada
from math import log
import ast # Importa o módulo ast para avaliar expressões

while True: # O "while True" executa o loop se o usuário quiser efetuar mais de um cálculo
    print(" Escolha a operação aritmética: 1- Operações Básicas 2- Exponenciação e Radiciação 3- Logaritmo") # Aqui mostra os tipos de operações que o usuário pode escolher executar
    tipo = int(input())
    if tipo == 1:
        print(" Escolha a operação: 1- Adição 2- Subtração 3-Multiplicação 4-Divisão ") # Aqui  mostra as operações básicas que o usuário pode escolher
        operação = int(input())
        if operação == 1: # Se for adição
            print(" Digite o primeiro número: ")
            n1 = float(input())
            print(" Digite o segundo número: ")
            n2 = float(input())
            resultado = n1 + n2
        else:
            if operação == 2: # Se for subtração
                print(" Digite o primeiro número: ")
                n1 = float(input())
                print(" Digite o segundo número: ")
                n2 = float(input())
                resultado = n1 - n2
            else:
                if operação == 3: # Se for multiplicação
                    print(" Digite o primeiro número: ")
                    n1 = float(input())
                    print(" Digite o segundo número: ")
                    n2 = float(input())
                    resultado = n1 * n2
                else: # Se for divisão
                    print(" Digite o primeiro número: ")
                    n1 = float(input())
                    print(" Digite o segundo número: ")
                    n2 = float(input())
                    if n2 == 0:
                        print("Erro: Não é possível dividir por zero.")# Tratamento de erro quando houver divisão por zero
                        continue  # Pula para a próxima iteração do loop
                    else:
                        resultado = n1 / n2
    else:
        if tipo == 2:
            print("Escolha a operação: 1-Exponenciação 2-Radicição ") # Irá apresentar as opções de potenciação e radicição
            operação = int(input())
            if operação == 1: # Se for exponenciação
                print(" Digite a base: ")
                n1 = float(input())
                print(" Digite o expoente: ")
                n2 = float(input())
                resultado =  n1 ** n2
            else: # Se for radiciação
                print(" Digite o índice: ")
                n1 = float(input())
                print(" Digite o radicando: ")
                n2 = float(input())
                resultado = n2 **( 1 / n1)
        else:
            if tipo == 3: # Irá para a opção do logaritmo
                print(" CONSIDERA-SE A BASE 10 ")
                print(" Digite o Logaritmando: ")
                n1 = float(input())
                if n1 <= 0:
                    print("Erro: O logaritmando deve ser maior que zero.")
                    continue # Pula para a próxima iteração do loop
                else:
                    resultado = log(n1) / log(10)

            else:
                print("inválido! ") # Se o usuário inserir qualquer número errado em relação a primeira lista será mostrada essa opção de tratameno de erro

    # Verifica se resultado foi definido antes de imprimir
    if 'resultado' in locals():
      print(" O resultado é: " + str(resultado))
      print("Resultado de último cálculo realizado")

    print(" Quer efetuar cálculos ainda? 1- SIM 2- SAIR") # Interação com o usuário para analisar se quer continuar efetuando cálculos ou sair
    resposta = int(input())
    if resposta != 1: break # Esse comando irá parar o programa se o usuário não quiser efetuar mais cálculos e resultará no termino do script
```

Este script de Python implementa uma calculadora simples de linha de comando que executa operações aritméticas básicas (adição, subtração, multiplicação, divisão), exponenciação, radiciação e logaritmo na base 10. Inclui tratamento de erros para divisão por zero e logaritmos de números não positivos. A calculadora permite ao usuário realizar vários cálculos sequencialmente até que ele escolha sair.
